```python
from bs4 import MarkupResemblesLocatorWarning
import warnings

import httpx

warnings.filterwarnings("ignore", category=MarkupResemblesLocatorWarning)
from bs4 import BeautifulSoup
search_url = "https://annas-archive.gl/search?index=&page=1&sort=&display=&q=IT%2BStephen+King%2Bepub"
async with httpx.AsyncClient(follow_redirects=True, timeout=15) as client:
    r = await client.get(search_url, headers={"User-Agent": "Mozilla/5.0"})
    r.raise_for_status()

soup = BeautifulSoup(r.content, "html.parser")
print(soup.prettify())

all_tags = soup.find_all(lambda tag: tag.name == 'a' and tag.text.strip().upper() == "IT")
if len(all_tags) == 0:
    all_tags = soup.find_all(lambda tag: tag.name == 'a' and  tag.text.strip().upper().startswith("IT"))
for tag in all_tags:
    print(tag.text.strip())
```
